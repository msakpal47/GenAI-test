```python
# # Bag of words
# ✅ Definition
# The Bag of Words model represents text (a sentence, document, etc.) as an unordered collection of words, disregarding grammar, order, and syntax, but keeping multiplicity (word frequency).


# ✅ How It Works
# Create Vocabulary: Extract all unique words from the text corpus.

# Count Occurrences: For each document, count how many times each word from the vocabulary appears.

# Represent each document as a vector of word counts (or sometimes TF-IDF scores).
```

```python
## Manual Implementation

documents = [
    "The food is good",
    "The food is bad",
    "The pizza is amazing",
    "The pasta tastes great",
    "I love spicy food",
    "The restaurant is clean",
    "Service was excellent today",
    "The waiter was rude",
    "The burger is delicious",
    "The fries were cold",
    "I enjoyed the dessert",
    "The soup was too salty",
    "The coffee tastes bitter",
    "The tea is refreshing",
    "The juice is fresh",
    "The price is reasonable",
    "The food was undercooked",
    "The meal was overcooked",
    "The salad is healthy",
    "The place is very noisy",
    "The environment is calm",
    "I had a great experience",
    "The staff was friendly",
    "The food was served late",
    "The food quality is poor",
    "I recommend this place",
    "I will never come back",
    "The pizza is too cheesy",
    "The ambiance is beautiful",
    "The menu has many options"
]
```

```python
documents
```

```python
#  Tokenize and create vocabulary
[doc.lower().split() for doc in documents]
```

```python
tokens = [doc.lower().split() for doc in documents]
```

```python
tokens
```

```python
sorted(set(word for sentence in tokens for word in sentence))
```

```python
vocab = sorted(set(word for sentence in tokens for word in sentence))
```

```python
vocab
```

```python
print("Vocabulary:",vocab)
```

```python
# Create Bag of Words representation
bow_matrix = []
for sentence in tokens:
    row = [sentence.count(word) for word in vocab]
    bow_matrix.append(row)
```

```python
# Display results
print("Vocabulary:",vocab)
print("Bag of words.matrix:")
for i,row in enumerate(bow_matrix):
    print(f"sentence{i+1}:",row)
```
