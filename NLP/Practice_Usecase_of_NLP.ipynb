```python
# Natural Language Toolkit

# The Natural Language Toolkit (NLTK) is a powerful, open-source Python library used for working with human language data (text) in the field of Natural Language Processing (NLP).

# It provides easy-to-use interfaces to:

# Text processing libraries

# Corpora (text datasets)

# Linguistic analysis tools

# Key Features of NLTK:
# ✅ Tokenization (word, sentence)
# ✅ Stopword removal
# ✅ Stemming and Lemmatization
# ✅ Part-of-speech (POS) tagging
# ✅ Named Entity Recognition (NER)
# ✅ Parsing and chunking
# ✅ WordNet interface (for meanings and synonyms)
# ✅ Text classification and sentiment analysis
# ✅ Access to large corpora like Brown, Gutenberg, etc.

```

```python
# spacy NLP


# Definition of spaCy in NLP:
# spaCy is a high-performance, open-source Python library designed for advanced Natural Language Processing (NLP). It's known for its speed, efficiency, and use in real-world applications (production-ready).

# Key Features of spaCy:
# ✅ Fast tokenization
# ✅ Part-of-Speech (POS) tagging
# ✅ Named Entity Recognition (NER)
# ✅ Dependency parsing
# ✅ Lemmatization
# ✅ Sentence segmentation
# ✅ Text classification
# ✅ Pre-trained models for multiple languages
# ✅ Integration with deep learning libraries (e.g., PyTorch, TensorFlow)



```

```python
# Difference Between Both

# Here’s a clear comparison between NLTK and spaCy, highlighting the major differences:

# Feature / Aspect	                NLTK (Natural Language Toolkit)	                                                  spaCy
# Purpose	                       Educational, research-focused	                                               Industrial, production-ready
# Speed	                      Slower (more flexible)	                                                        Faster and more efficient
# Ease of Use	                 Good for beginners, but APIs can be verbose	                                   Clean, concise, modern API
# Installation Size	          Lightweight with modular downloads	                                           Heavier due to pre-trained models
# Pre-trained Models	         Limited, requires manual setup	                                               Built-in, high-quality pre-trained models
# Tokenization	                   Basic, customizable	                                                           Very fast, accurate, and built-in
# POS Tagging	                               Available (not context-based)	                                               Fast and context-aware tagging
# Named Entity Recognition (NER)	              Available, needs more setup              	                           Built-in, accurate, fast
# Lemmatization	                            Available (uses WordNet)	                                                   More accurate, built-in
# Text Classification	                     Manual implementation	                                                        Built-in pipeline or customizable
# Deep Learning Support	                    Limited	                                                                   Strong support via spaCy Transformers
# Corpus Access	                        Includes many corpora like Brown, Gutenberg	                                   Limited corpus access
# Customization	                     Highly customizable for learning & research	                                       Custom pipelines and training possible
# Use Case Fit	                        Best for learning, research, and prototyping	                                   Best for production and deployment

# ✅ Summary:
# Use NLTK if you're:

# A beginner learning NLP concepts

# Doing research or working with text corpora

# Building small-scale, educational projects

# Use spaCy if you're:

# Building real-world, production-grade NLP applications

# Need performance and speed

# Want modern NLP tools with pre-trained models

```

```python
# Stemming in NLP
# Stemming is the process of reducing a word to its root or base form (called a stem) by removing suffixes and prefixes.

# 💡 It helps normalize words in text processing so that similar words are treated the same (e.g., "running", "runs", and "runner" → "run").
```

```python
# ✅ Porter Stemmer – Definition:
# The Porter Stemmer is a popular stemming algorithm in Natural Language Processing (NLP) that reduces words to their root form (called a "stem"). Developed by Martin Porter in 1980, it uses a series of rules to remove common morphological and inflectional endings from English words.





from nltk.stem import PorterStemmer

PortStemmer = PorterStemmer()
```

```python
words = [
    "run", "running", "ran", "runs",
    "eat", "eating", "eaten", "eats",
    "go", "going", "gone", "goes",
    "do", "doing", "done", "does",
    "be", "is", "are", "was", "were", "being", "been",
    "have", "has", "had", "having",
    "fly", "flying", "flew", "flown",
    "write", "writing", "wrote", "written", "writes",
    "read", "reading", "reads",
    "drink", "drinking", "drank", "drunk", "drinks",
    "sleep", "sleeping", "slept", "sleeps",
    "speak", "speaking", "spoke", "spoken", "speaks",
    "break", "breaking", "broke", "broken", "breaks",
    "buy", "buying", "bought", "buys",
    "sell", "selling", "sold", "sells",
    "know", "knowing", "knew", "known", "knows",
    "think", "thinking", "thought", "thinks",
    "swim", "swimming", "swam", "swum", "swims",
    "build", "building", "built", "builds",
    "sing", "singing", "sang", "sung", "sings",
    "teach", "teaching", "taught", "teaches",
    "take", "taking", "took", "taken", "takes"
]
```

```python
for word in words:
    print(word + "------>" + PortStemmer.stem(word))
```

```python
#  Definition of RegexpStemmer Class
# The RegexpStemmer is a class in the NLTK library that performs stemming based on custom regular expression (regex) rules you define. 
# Unlike rule-based stemmers like Porter, this stemmer allows you to create your own patterns to strip suffixes or prefixes by matching regex patterns.

# What is RegexpStemmer?
# It removes the longest matching suffix defined by your regex from the word.

# You provide a regex pattern describing which suffixes to remove.

# Useful when you want custom, lightweight stemming or suffix removal.


# Parameters
# regexp : str or regexp
# The regular expression that should be used to identify morphological affixes.

# min : int
# The minimum length of string to stem

from nltk.stem import RegexpStemmer

Stemmer = RegexpStemmer('ing$|ed$|es$', min=1)

# List of words for testing
wording = [
    "playing",
    "tables",
    "danceable",
    "agrees",
    "bake",
    "riding",
    "laughs",
    "capable",
    "creates",
    "working"
]

```

```python
for word in wording:
    print(f"{word} ----> {Stemmer.stem(word)}")
```

```python
from nltk.stem import RegexpStemmer

stemmer = RegexpStemmer('ing$|s$', min=1)
print(stemmer.stem('is'))  #
```

```python
stemmer = RegexpStemmer('ing$|s$', min=2)
print(stemmer.stem('is'))  # Output: 'is' (unchanged, stem would be too short)
```

```python
stemmer = RegexpStemmer('ing$', min=3)

print(stemmer.stem('playing'))  # play (length 4, OK)
print(stemmer.stem('sing'))     # sing (would become 's', which is < 3, so unchanged)
```
